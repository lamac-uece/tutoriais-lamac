### O que é Numpy?

Numpy é uma biblioteca matemática robusta de python, amplamente utilizada nas áreas de análise de dados, ciência de dados, machine learning, entre outras. Com ela é possível realizar operações complexas e manipulações de dados numéricos de maneira mais eficiente do que com os métodos "padrões" da linguagem python.

### Fazendo a importação

Por conveção da comunidade usa-se `as np`.

```python
import numpy as np
```

### Numpy Array

Uma estrutura de dados da biblioteca numpy, usada para armazenar coleções ordenadas de dados. Os arrays do numpy oferecem um armazenamento mais eficiente e um processamento mais rápido do que o tipo `list`, além de possuirem métodos numéricos, estatísticos e de álgebra linear.

```python
lista = [1, 2, 3, 4, 5]

arr = np.array(lista)

print(type(lista))
print(type(arr))
print("Somatório dos números da lista:", arr.sum())
print("Maior número da lista:", arr.max())
print("Menor número da lista:", arr.min())
print("Média dos números da lista:", arr.mean())
```

Os arrays do numpy podem ser tanto unidimensionais como também multidimensionais, com arrays bidimensionais podendo ser interpretados como matrizes. O atributo _shape_ indica o número de dimensões e a magnitude de cada uma (no caso de um array 2d, isso pode ser visto como o número de linhas e o número de colunas).

```python
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]
lista3 = [7, 8, 9]

matrix = np.array([lista1, lista2, lista3])

print(type(matrix))
print("Número de linhas (n) e de colunas (m) da matriz: (n, m) =", matrix.shape)
print("Diagonal principal da matriz:", matrix.diagonal())
```

### Salvando e carregando arrays

O numpy tem métodos para salvar arrays como arquivos de extensão `.npy`, bem como carregar esses arquivos em variáveis do tipo array, preservando os dados e a estrutura. Essa funcionalidade é bastante útil para a reutilização e prevenção a perda dos dados.

```python
# Cria uma matriz de zeros 3x5 
matrix = np.zeros((3, 5))

print(matrix)
print(matrix.shape)

np.save('arquivo.npy', matrix)
```

Carregando o arquivo

```python
new_matrix = np.load('arquivo.npy')

print(new_matrix)
print(new_matrix.shape)
```

### Numpy random

Um módulo da biblioteca numpy que possui métodos para manipulação de números aleatórios. É particularmente útil para trabalhar com operações probabilísticas.

É possível gerar números aleatórios

```python
random_num = np.random.randint(1, 20)

random_matrix = np.random.rand(3,4)

print('Número inteiro aleatório entre 1 e 20:', random_num)

print('Matriz de números aleatórios entre 0 e 1:\n', random_matrix)
```

Também é possível selecionar um item aleatoriamente de um conjunto de dados

```python
persons = np.array(['Joaquim', 'Maria', 'Pedro', 'Paulo', 'Antônia', 'Larissa'])

chosen = np.random.choice(persons)

print('A pessoa sorteada foi:', chosen)
```
